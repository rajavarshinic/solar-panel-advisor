```python
def recommend_solar_placement(df, latitude):
    # Filter data based on time of day (Let's consider daytime from 6 AM to 6 PM)
    daytime = df[(df['DATE_TIME'].str[11:13].astype(int) >= 6) & (df['DATE_TIME'].str[11:13].astype(int) <= 18)]

    # Find the highest solar irradiation
    best_time = daytime.loc[daytime['IRRADIATION'].idxmax()]  # Find row with highest irradiation

    # Latitude and weather-based suggestions
    print(f"🌞 Best Time for Solar Panel Placement:")
    print(f"📅 Hour: {best_time['DATE_TIME']}")
    print(f"🌡 Temperature: {best_time['AMBIENT_TEMPERATURE']}°C")
    print(f"🔆 Solar Irradiation: {best_time['IRRADIATION']} W/m²")
    print(f"🌍 Suggested Latitude: {latitude}°")

    print("\n🚀 Conclusion: Solar panels should be placed at optimal angles between 6 AM - 6 PM with maximum irradiation.")
```

```python
# Example: Let's check recommendations for latitude 13° (Chennai)
recommend_solar_placement(df, 13)
```
